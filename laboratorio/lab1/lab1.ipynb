# Determinación de los valores de b y w de una línea recta

## Objetivo

El objetivo es encontrar los mejores valores de `b` y `w` para la ecuación:

y = b + wx

Se utilizará el dataset Salary, donde:

- `X` = YearsExperience (años de experiencia)
- `Y` = Salary (salario)

Se utilizarán los 35 registros del dataset para calcular los valores de `b` y `w`.

```python
import numpy as np
data = np.loadtxt(
    'salary.txt',
    delimiter=',',
    skiprows=1
)
print(data)
```

La primera columna corresponde a YearsExperience, por lo que se utiliza como X. La segunda columna corresponde a Salary, por lo que se utiliza como Y.

Se obtiene la cantidad de registros del dataset. En este caso se utilizan 35 datos.

```python
X = data[:, 0]
y=data[:, 1]
n=len(X)
print("X:", X)
print("Y:", y)
print(n)
```

Se crean variables que comenzarán en cero y servirán para acumular las sumatorias necesarias para calcular w y b

```python
suma_x = 0
suma_y = 0
suma_xy = 0
suma_x2 = 0
```

Se utiliza un ciclo for para recorrer los 35 registros y calcular las sumatorias de X, Y, X*Y y X²

```python
for i in range(n):

    suma_x = suma_x + X[i]

    suma_y = suma_y + y[i]

    suma_xy = suma_xy + (X[i] * y[i])

    suma_x2 = suma_x2 + (X[i] * X[i])
```

Se calcula w, que representa la pendiente de la línea. El resultado obtenido fue aproximadamente:

```python
numerador = (n * suma_xy) - (suma_x * suma_y)

denominador = (n * suma_x2) - (suma_x * suma_x)

w = numerador / denominador

print("Valor de w:", w)
```

Esto significa que el salario aumenta aproximadamente 8731.94 por cada año adicional de experiencia, según el modelo.

Se calcula b, que representa el punto donde la línea corta el eje Y

```python
promedio_x = suma_x / n

promedio_y = suma_y / n

b = promedio_y - (w * promedio_x)

print("Valor de b:", b)
```

Se utiliza nuevamente un for para calcular el salario estimado de cada uno de los 35 registros utilizando la ecuación obtenida.

```python
print("Predicciones:")
print()

for i in range(n):

    prediccion = b + (w * X[i])

    print(
        "x =", X[i],
        "Real =", y[i],
        "Predicción =", round(prediccion, 2)
    )
```

Se crean 10 nuevos valores de años de experiencia

```python
nuevos_x = [
    1.2,
    1.7,
    2.5,
    3.5,
    4.3,
    5.5,
    6.5,
    7.5,
    8.5,
    10.0
]
```

Se utiliza la ecuación encontrada para estimar el salario correspondiente a cada uno de los 10 nuevos valores de experiencia.

```python
for i in range(len(nuevos_x)):

    nuevo_y = b + (w * nuevos_x[i])

    print(
        "Años de experiencia:", nuevos_x[i],
        "Salario estimado:", round(nuevo_y, 2)
    )
```

```python
if w > 0:

    print("La relación entre experiencia y salario es positiva.")

elif w < 0:

    print("La relación entre experiencia y salario es negativa.")

else:

    print("No existe una relación lineal.")
```

Se utiliza una estructura decisiva if, elif y else para determinar el tipo de relación entre las variables.

Como w es positivo, el programa indica que existe una relación positiva: a mayor experiencia, mayor salario estimado.
